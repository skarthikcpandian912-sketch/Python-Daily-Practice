```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Machine": ["M001", "M002", "M003", "M004", "M005",
                "M006", "M007", "M008", "M009", "M010"],
    "Temperature": [45, 55, 65, 75, 85, 50, 60, 70, 90, None],
    "Running_Hours": [2, 4, 6, 8, 10, 3, 5, 7, 12, 9],
    "Vibration": [1.2, 1.5, 2.0, 2.8, 3.5, 1.3, 1.8, 2.5, 4.0, 3.0]
}

df = pd.DataFrame(data)

print(df)
```

```python
print(df.isnull().sum())
```

```python
print(df.duplicated().sum())
```

```python
print(df["Temperature"].describe())
```

```python
correlation = df["Running_Hours"].corr(df["Temperature"])

print("Correlation:", correlation)
```

## EDA Observations and Insights

### Observations
1. The dataset contains 10 machine records.
2. The Temperature column contains 1 missing value.
3. No duplicate records were found.
4. The maximum recorded temperature is 90°C.
5. Running Hours and Temperature have a strong positive correlation of approximately 0.99.

### Insights
1. Machines operating for longer hours tend to have higher temperatures in this dataset.
2. Temperature and running hours may be useful features for a predictive maintenance model.
3. The missing temperature value should be investigated or handled before model training.

### Conclusion
The EDA identified a strong positive relationship between machine running hours and temperature. Further analysis with more machine data is needed to understand the causes of overheating.
